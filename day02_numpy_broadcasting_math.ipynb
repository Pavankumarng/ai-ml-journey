```python
import numpy as np
```

```python
a = np.array([[1],
              [2],
              [3]])

b = np.array([[10, 20, 30, 40]])

print(a.shape)
print(b.shape)

print(a + b)
```

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[10, 20],
              [30, 40]])

print(a * b)
```

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[10, 20],
              [30, 40]])

print(np.dot(a, b))
```

```python
x = np.array([2, 4, 6, 8, 10])

print("sum:", np.sum(x))
print("mean:", np.mean(x))
print("std:", np.std(x))
print("var:", np.var(x))
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
result = 0

for i in range(len(a)):
    result += a[i] * b[i]

print(result)
```

```python
print(np.dot(a, b))
```

```python
x = np.array([10, 20, 30, 40, 50])
print(x[[0, 2, 4]])
```

```python
x = np.array([10, 20, 30, 40, 50])
print(x > 25)
print(x[x > 25])
```
