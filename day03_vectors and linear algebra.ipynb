```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([3, 1])
B = np.array([2, 2])
C = A + B

# Vector A
plt.quiver(0, 0, A[0], A[1],
           angles='xy', scale_units='xy', scale=1)

# Vector B starting from the end of A
plt.quiver(A[0], A[1], B[0], B[1],
           angles='xy', scale_units='xy', scale=1)

# Result A + B
plt.quiver(0, 0, C[0], C[1],
           angles='xy', scale_units='xy', scale=1)

plt.xlim(0, 6)
plt.ylim(0, 5)
plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Geometric Verification of A + B")

plt.show()
```
