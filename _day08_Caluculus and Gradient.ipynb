```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def loss(w):
    return w**2
print("Loss when w = 3:", loss(3))
print("Loss when w = 2:", loss(2))
print("Loss when w = 0:", loss(0))
```

```python
def gradient(w):
    return 2 * w
print("Gradient when w = 3:", gradient(3))
print("Gradient when w = 2:", gradient(2))
print("Gradient when w = -3:", gradient(-3))
```

```python
w = 3.0
learning_rate = 0.1

grad = gradient(w)

new_w = w - learning_rate * grad

print("Old w:", w)
print("Gradient:", grad)
print("Learning rate:", learning_rate)
print("New w:", new_w)
```

```python
old_loss = loss(w)
new_loss = loss(new_w)

print("Old loss:", old_loss)
print("New loss:", new_loss)
```

```python
w = 3.0
learning_rate = 0.1

for i in range(10):
    grad = gradient(w)
    w = w - learning_rate * grad

    print("Step:", i + 1, "| w:", w, "| Loss:", loss(w))
```

```python
def f(x):
    return x**2 + 3*x + 1
print(f(2))
```

```python
def f(x):
    return x**2 + 3*x + 1

print(f(2))
```

```python
x0 = 2
h = 0.001

print("f(x0) =", f(x0))
print("f(x0 + h) =", f(x0 + h))
```

```python
derivative = (f(x0 + h) - f(x0)) / h

print("Numerical derivative =", derivative)
```

```python
x0 = 2
y0 = f(x0)

slope = derivative

print("Point:", (x0, y0))
print("Slope:", slope)
```

```python
import numpy as np
```

```python
x = np.linspace(-2, 6, 400)

tangent = slope * (x - x0) + y0

print("First 5 x values:", x[:5])
print("First 5 tangent values:", tangent[:5])
```

```python
y = f(x)

print("First 5 function values:", y[:5])
print("First 5 tangent values:", tangent[:5])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x, y, label="Original function")
plt.plot(x, tangent, label="Tangent line")

plt.scatter(x0, y0, label="Point (2, 11)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Function and Tangent Line")
plt.legend()
plt.grid(True)

plt.show()
```
