```python
import numpy as np

# 1. Training data
X = np.array([1000, 1500, 2000, 2500])
y = np.array([40, 60, 80, 100])

# 2. Add a column of 1s for the intercept
X_b = np.c_[np.ones(len(X)), X]

# 3. Normal Equation
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

# 4. Extract parameters
b = theta[0]
m = theta[1]

print("Intercept:", b) 
print("Slope:", m)

# 5. Make predictions
predictions = X_b @ theta

print("Predictions:", predictions)

# 6. Calculate MSE
mse = np.mean((y - predictions) ** 2)

print("MSE:", mse)

# 7. Calculate R²
sse = np.sum((y - predictions) ** 2)
sst = np.sum((y - np.mean(y)) ** 2)

r2 = 1 - (sse / sst)

print("R²:", r2)

# 8. Predict a new house
new_house = np.array([[1, 3000]])

new_prediction = new_house @ theta

print("Predicted price for 3000 sq ft:",
      new_prediction[0])
```

```python
import numpy as np
import pandas as pd


# 1. Load the dataset
df = pd.read_csv("real_estate_dataset.csv")


# 2. Select input and target
X = df["Square_Feet"].values
y = df["Price"].values


# 3. Add a column of 1s for the intercept
X_b = np.c_[np.ones(len(X)), X]


# 4. Calculate theta using Normal Equation
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y


# 5. Get intercept and slope
b = theta[0]
m = theta[1]


# 6. Make predictions
predictions = X_b @ theta


# 7. Calculate MSE
mse = np.mean((y - predictions) ** 2)


# 8. Calculate R²
sse = np.sum((y - predictions) ** 2)
sst = np.sum((y - np.mean(y)) ** 2)

r2 = 1 - (sse / sst)


# 9. Display results
print("Intercept (b):", b)
print("Slope (m):", m)
print("MSE:", mse)
print("R²:", r2)


# 10. Predict price for a new house
new_house = np.array([[1, 10000]])

new_prediction = new_house @ theta

print("Predicted price for 10000 square feet:", new_prediction[0])
```
