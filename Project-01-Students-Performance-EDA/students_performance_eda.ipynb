```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("StudentsPerformance.csv")
```

```python
sns.boxplot(data=df, y="math score")

plt.title("Boxplot of Mathematics Scores")
plt.ylabel("Math Score")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df[["math score", "reading score", "writing score"]])

plt.title("Comparison of Scores Across Subjects")
plt.ylabel("Marks")

plt.show()
```

```python
corr = df[["math score", "reading score", "writing score"]].corr()

print(corr)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)

plt.title("Correlation Between Exam Scores")
plt.show()
```

```python
df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean()
```

```python
df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]
].mean()
```

```python
prep_scores = df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean()

prep_scores.plot(kind="bar", figsize=(9, 5))

plt.title("Average Scores by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.show()
```

```python
parent_scores = df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]
].mean()

parent_scores.plot(kind="bar", figsize=(12, 6))

plt.title("Average Scores by Parental Education")
plt.xlabel("Parental Education Level")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

## Key Findings and Observations

After exploring the dataset and comparing students' performance across different subjects, I noticed a few interesting patterns:
1. Overall subject performance
Reading had the highest average score at 69.17, followed by Writing at 68.05. Mathematics had the lowest average at 66.09. This shows that students performed slightly better in Reading and Writing compared to Mathematics.
2. Relationship between subjects
Reading and Writing scores showed a very strong positive correlation of 0.95. Mathematics also had a strong positive relationship with Reading (0.82) and Writing (0.80). This suggests that students who performed well in one subject generally tended to perform well in the others.
3. Impact of test preparation
Students who completed the test preparation course had higher average scores in all three subjects compared to those who did not. The difference was particularly noticeable in Writing, where the average score was nearly 10 marks higher.
4. Parental education and performance
Students whose parents had a master's degree recorded the highest average scores in Mathematics, Reading and Writing. Students whose parents had completed only high school had the lowest averages across all three subjects.
5. Variation in subject scores
Writing scores showed the highest variation, with a standard deviation of 15.20, while Reading scores had the lowest variation at 14.60. This means Writing scores were slightly more spread out than Reading scores.
